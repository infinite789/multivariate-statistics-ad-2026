```python
import numpy as np
from pathlib import Path
import matplotlib.pyplot as plt

def true_function(x):
    return np.sin(2 * np.pi * x)

def plot_overfitting(output_path="beamer/images/overfitting.png", random_seed=42):
    rng = np.random.default_rng(random_seed)

    # Малка извадка с шум
    x = np.sort(rng.uniform(-1, 1, 10))
    y = true_function(x) + rng.normal(0, 0.6, size=x.size)

    # Полином от 15-та степен: 16 коефициента за 16 точки
    model = np.polynomial.Polynomial.fit(x, y, deg=10)

    x_plot = np.linspace(-1, 1, 1000)
    y_plot = model(x_plot)

    plt.figure(figsize=(10, 6))
    plt.scatter(x, y, color="red", s=55, label="Наблюдавани данни", zorder=3)
    plt.plot(x_plot, y_plot, color="blue", label="Полином от 20-ти ред")
    plt.plot(
        x_plot, true_function(x_plot), "--",
        color="green", label="Истинска функция"
    )

    plt.title("Overfitting")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.xlim(-1, 1)
    plt.ylim(-5, 5)
    plt.grid(True, alpha=0.3)
    plt.legend()

    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)
    plt.savefig(output_path, dpi=300, bbox_inches="tight")
    plt.show()

    return model
```

```python
model = plot_overfitting(output_path="beamer/images/overfitting_1.png", random_seed=2)
```

```python
model = plot_overfitting(output_path="beamer/images/overfitting_2.png", random_seed=3)
```

```python
def plot_underfitting(output_path="beamer/images/underfitting.png", random_seed=42):
    rng = np.random.default_rng(random_seed)

    x = np.sort(rng.uniform(-1, 1, 10))
    y = true_function(x) + rng.normal(0, 0.6, size=x.size)

    # Линеен модел: недообучен модел
    model = np.polynomial.Polynomial.fit(x, y, deg=1)

    x_plot = np.linspace(-1, 1, 1000)
    y_plot = model(x_plot)

    plt.figure(figsize=(10, 6))
    plt.scatter(x, y, color="red", s=55, label="Наблюдавани данни", zorder=3)
    plt.plot(x_plot, y_plot, color="blue", label="Линейна регресия")
    plt.plot(
        x_plot,
        true_function(x_plot),
        "--",
        color="green",
        label="Истинска функция"
    )

    plt.title("Underfitting")
    plt.xlabel("x")
    plt.ylabel("y")
    plt.xlim(-1, 1)
    plt.ylim(-5, 5)
    plt.grid(True, alpha=0.3)
    plt.legend()

    output_path = Path(output_path)
    output_path.parent.mkdir(parents=True, exist_ok=True)
    plt.savefig(output_path, dpi=300, bbox_inches="tight")
    plt.show()

    return model
```

```python
model = plot_underfitting(
    output_path="beamer/images/underfitting_1.png",
    random_seed=42
)
```
