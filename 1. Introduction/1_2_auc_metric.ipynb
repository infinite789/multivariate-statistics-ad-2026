```python
import numpy as np
from sklearn.metrics import auc

import matplotlib.pyplot as plt

# Хипотетични ROC криви
fpr = np.array([0.0, 0.05, 0.10, 0.20, 0.35, 0.50, 0.70, 1.0])

tpr_model_1 = np.array([0.0, 0.45, 0.65, 0.78, 0.87, 0.92, 0.97, 1.0])
tpr_model_2 = np.array([0.0, 0.30, 0.50, 0.68, 0.78, 0.86, 0.94, 1.0])
tpr_model_3 = np.array([0.0, 0.20, 0.38, 0.55, 0.68, 0.77, 0.88, 1.0])

auc_1 = auc(fpr, tpr_model_1)
auc_2 = auc(fpr, tpr_model_2)
auc_3 = auc(fpr, tpr_model_3)

plt.figure(figsize=(8, 6))

plt.plot(fpr, tpr_model_1, linewidth=2, label=f"Модел 1 (AUC = {auc_1:.2f})")
plt.plot(fpr, tpr_model_2, linewidth=2, label=f"Модел 2 (AUC = {auc_2:.2f})")
plt.plot(fpr, tpr_model_3, linewidth=2, label=f"Модел 3 (AUC = {auc_3:.2f})")
plt.plot([0, 1], [0, 1], "k--", label="Случаен модел (AUC = 0.50)")

plt.xlabel("False Positive Rate (1 - Specificity)", fontsize=12)
plt.ylabel("True Positive Rate (Sensitivity)", fontsize=12)
plt.title("ROC - AUC", fontsize=14)
plt.xlim(0, 1)
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.legend(loc="lower right")
plt.show()
```
