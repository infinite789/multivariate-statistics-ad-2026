### 1.3.8 Цикли (for, while, break, continue)
#### for cycle
С for цикъла можем да преминем през всеки елемент на една колекция и да изпълним блок от код с този елемент като променлива

````python
arr = [1, 2, 'foo', 'asdf', 5.3, -12.4, 34, 12.9, 'hello', 1, 2, 3, 'appended']
for item in arr:
    print(item)
````

````python
s = 0
text = ''
for item in arr:
    if isinstance(item, str):
        text += item + ' '
    else:
        s += item
print(s)
print(text)
````

````python
sum_of_squares = 0
for i in range(-100, 100):
    sum_of_squares += i ** 2
print(sum_of_squares)
````

Тук сме изпозлвали вградената функция range. Тя ни позволява да итерираме (да преминем през) всички числа в определен интервал с определена стъпка (тук не сме задали стъпка, затова тя се приема да е 1)

#### while cycle
С while цикъл можем да повтаряме даден блок от код докато определена логическа проверка е True.

````python
txt = ''
word = 'hey'
while len(txt) < 120:
    txt += word + ' '
print(txt)
````

Тук сме използвали вградената функция len, която ни дава дължината на един обект (ако той има такава - len(5) ще хвърли грешка, понеже целите числа нямат дължина).

Пример за пресмятане на хармонично число: $s = \sum_{n=1}^{99} \frac{1}{n}$.

````python
s = 0
n = 1
while n < 100:
    s += 1 / n
    n += 1
print(s)
````

#### break
В горния пример сумираме 1/n за числата до 100. Ако сместо това искаме да сумираме, докато вече s не се променя значително, т.е. 1/n е по-малко от някакъв епсилон eps, можем да използваме break. Така можем да прекъснем цикъла отвътре, ако дадено условие е изпълнено. (разбира се можем просто да сметнем колко трябва да е n за определено eps, но не се сетих за по-добър пример за break)

````python
eps = 3.5e-7 # scientific notation 3.5 * 10^(-7)
s = 0
n = 1
while True: # този цикъл ще върви завинаги, ако не излезем от него с break
    one_over_n = 1/n
    if one_over_n <= eps:
        # спираме цикъла
        break
    s += one_over_n
    n += 1
print(f"The sum of 1/n for n from 1 to {n - 1} is {s:.2f}") # с :.2f опказваме само 2 знака след запетаята
````

#### continue

С continue можем да прескочим едно завъртане на цикъла. Ако в горния пример искаме да сумираме 1/n без тези n, които се делят на 7, можем да го направим с continue

````python
s = 0
n = 1
while True:
    if not n % 7:
        n += 1
        continue
    one_over_n = 1/n
    if one_over_n <= eps:
        break
    s += one_over_n
    n += 1
print(f"The sum of 1/n for n from 1 to {n - 1} excluding multiples of 7 is {s:.2f}") # с :.2f опказваме само 2 знака след запетаята
````

Тук n % 7 връща 0, ако n се дели на 7, което като bool се счита за False. Обръщаме го с not пред него. Така ако n се делина на 7, просто увеличаваме n с едно и се връщаме в началото на цикъла.

Когато стигнем до **return**, излизаме от функцията и останалата част от кода в нея не се изълнява. Ако не искаме резултатът от последната функция да става прекалено дълъг, можем да поставим условие за string и n, което ако не е изпълнено, да излизаме от функцията (не връщаме нищо).

````python
def copy_string(string, n):
    if n * len(string) > 100:
        print('The result would be too long!')
        return
    # можем да "умножаваме" стрингове с цели числа, което просто повтаря стринга n пъти
    return string * n
````

````python
copied_string = copy_string('hello', 7)
print(copied_string)
````

````python
copied_string = copy_string('hello', 70)
print(copied_string) # None is a special value for "nothing"
````

Когато дефинираме функцията, можем да задаваме стойности по подразбиране на аргументите и така не е нужно да ги подаваме, когато викаме функцията

````python
def copy_string(string, n=10):
    if n * len(string) > 100:
        print('The result would be too long!')
        return
    return string * n
````

````python
# а thousand normally distributed samples with mean 5 and sigma 2
n = 1000
samples = [random.gauss(5, 2) for _ in range(n)]
total = sum(samples)
mean = total / n
print(f"Sample mean = {mean}")
print(f"Sample var = {math.sqrt(sum((s - mean) ** 2 for s in samples) / n)}")
````

numpy ни позволява да индексираме масиви с друг масив от boolean стоиности със същата размерност. Така можем да филтрираме (маскираме) част от данните:

````python
arr = np.array([1,2,5,2,35,643,23,6234,235,124])
odd = (arr % 2).astype(bool)
print(odd)
print(arr[odd])
print(arr[~odd]) # negation
print(arr[odd & (arr > 100)])
````

matplotlib е библитека за чертаене на графики. seaborn е библиотке, която използва matplotlib и имлпементира грефики, специфични за статистическия анализ. Както винаги, първо ги инсталираме:

Няма нужда да инсталираме matplotlib експлицитно, понеже seaborn изисква matplotlib да бъде инсталиран, за да работи. Когато инсталираме seaborn, pip разбира това и инсталира и matplotlib. След това можем да импортнем частта от matpllotlib, която ни позволява да го ползваме от python:

````python
import matplotlib.pyplot as plt
````

Тук виждаме, че можем да импортваме само специфични части от библиотеките. Не ни трябват всички функции от matplotlib, затова импортваме само pyplot частта.

````python
x = [i for i in range(100)]
y = [i ** 2 for i in x]
````

````python
plt.plot(x, y)
````

````python
points = np.random.multivariate_normal([13, 7], [[2, 6], [6, 20]], size=1000) # 2d gauss
points
````

````python
x = points.T[0]
y = points.T[1]
plt.scatter(x, y, color='g', label='random points')
plt.legend()
````

matplotlib има широка функционалност, но за по-сложни графики иска доста писане и експлицитно задаване на параметри. Затова е много по-лесно да изпозлваме seaborn, където голяма част от този допълнителен год е вече написан от някого другиго. Например, ако искаме да направим графика на линейна регресия, това става с един ред на seaborn:

````python
data = pd.DataFrame(points, columns=['x', 'y']) # така създаваме pandas.DataFrame от numpy масиви (можем и направо от обикновенни python листове)
data
````

````python
import seaborn as sns
````

````python
sns.lmplot(x='x', y='y', data=data)
````

Виждаме, че начина на подаване на данните към seaborn прилича много повече на този в R (подаваме само имената на колоните и отделно DataFrame с данните, от които да се вземат тези колони) отколкото този при matplotlib. Понеже данните са генерирани от гаусово разпределение, имаме почти перфектен фит на правата. Ако вкараме допълнителна грашка, ще видим, че seaborn я репрезентира автоматично без ние да правим нищо допълнително

````python
data['y'] += 70 * np.random.random(len(data))
````

````python
sns.lmplot(x='x', y='y', data=data, palette=['red'])
````

Това е само графична репрезентация на линеен фит на точките. За да свикнем с изпозлването на python за статистически анализ, в следващото упражнение ще разгледаме проста линейна регресия чрез библитеката statsmodels.
